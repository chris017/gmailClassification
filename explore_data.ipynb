# Gmail Spam Data — Exploration

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("emails.db")
df = pd.read_sql("SELECT * FROM emails", conn)
conn.close()

print(f"Total emails: {len(df)}")
print(f"Columns: {list(df.columns)}")
```

## Label Distribution

```python
df["label"].value_counts()
```

## Erste Zeilen

```python
df.head(10)
```

## Numerische Features nach Label

```python
df.groupby("label")[["body_length", "link_count", "caps_ratio"]].mean().round(3)
```

## Top Spam-Sender Domains

```python
df[df["label"] == "spam"]["sender_domain"].value_counts().head(15)
```

## Beispiel Spam-Email

```python
sample = df[df["label"] == "spam"].iloc[0]
print(f"From:    {sample['sender']}")
print(f"Subject: {sample['subject']}")
print(f"Links:   {sample['link_count']}")
print(f"Caps:    {sample['caps_ratio']:.1%}")
print(f"\nBody preview:\n{sample['body_preview']}")
```
